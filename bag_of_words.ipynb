```python
questions = [
    "What is the capital of France?",
    "Where is the Eiffel Tower located?",
    "How to bake a chocolate cake?"
]

stop_words=["is","the","a"]
final_s=""
```

```python
for i in questions:
  doc=i.split(" ")
  # for j in range(len(doc)):
  #   if doc[j] in stop_words:
  #     doc[j]=""
  for k in doc:
    if k=="":
      continue
    else:
      final_s+=k+" "

import re
final_s= re.sub(r"[?]","",final_s)

final_s=final_s.split(" ")
final_s.remove("")
```

```python
final_s
```

```python
def make_vector(doc,final_s):
  vec=[0]*len(final_s)
  for word in doc.split(" "):
    if word in final_s:
      vec[final_s.index(word)]+=1
  return vec
```

```python
final_v=[]
```

```python
for doc in questions:
  final_v.append(make_vector(doc,final_s))
```

```python
final_v
```

```python
def cosine_similarity(v1,v2):
  dot = sum(a*b for a,b in zip(v1,v2))

  mag1=sum(a**2 for a in v1) ** 0.5
  mag2=sum(b**2 for b in v2) ** 0.5

  return dot/(mag1*mag2)
```
